```python
# convert numpy into tensors and vice versa

# Numpy is a popular scitentifc Python numerical computing library. And due to this PyTorch needs intercation with numpy array.

array = np.arange(1.0, 8.0)
array

tensor = torch.from_numpy(array) # convert numpy array into tensor
print(f" tensor: {tensor} array: {array}") # pytorch is going to retain the default data type defined from numpy array which float64 unless specified otherwise


# default data type in numpy array is 'float64'
print(f"default data type of numpy array is: {array.dtype}")

# default data type in pytorch is 'float32'
my_torch = torch.ones(7)
print(f"default data type of pytorch is: {my_torch.dtype}")


# change the data type while converting numpy array into tensor
array_1  = np.arange(11.0, 20.0)
tensor_1 = torch.from_numpy(array_1).type(torch.float32)
print(f" tensor: {tensor_1}; tensor data type: {tensor_1.dtype} array: {array_1}; array data type: {array_1.dtype}")


# changing  the value of array doesn't change the value of the tensor
array_1 = array_1 + 1
print(f"array_1 : {array_1}")
print(f" tensor: {tensor_1}")

# convert tensor into numpy array - and
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
print(f"Numpy array: {numpy_tensor}; data type: {numpy_tensor.dtype}") # data type in converted array would be set to the original data type set for tensor which is float32 in this case

# change the value of tensor
tensor = tensor + 1

print(f"numpy array: {numpy_tensor}")
```
