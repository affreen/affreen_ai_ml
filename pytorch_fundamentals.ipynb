```python
import torch
```

```python
scaler = torch.tensor(7)
scaler

# get dimension
print(f"Dimension of scaler is {scaler.ndim}")

# get element from tensor
print(f"Element from scalaer is {scaler.item()}")
```

```python
vector = torch.tensor([7,7])
# get dimension
print(f"Dimension of vector is {vector.ndim} and shape is {vector.shape}")
```

```python
MATRIX = torch.tensor([[7,8],
                       [9,10]])
print(f"Dimension of MATRIX is {MATRIX.ndim} and shape is {MATRIX.shape}")

MATRIX_1 = torch.tensor([[7,8],
                       [9,10],
                       [11,12]])
print(f"Dimension of MATRIX_1 is {MATRIX_1.ndim} and shape is {MATRIX_1.shape}")

MATRIX_2 = torch.tensor([[[5,6],
                          [7,8],
                          [0,8]]])
```

```python
TENSOR = torch.tensor([[[[1,2,3],
                        [3,6,9],
                        [2,4,5]],
                       [[1,2,3],
                        [3,6,9],
                        [2,4,5]],
                       ]])

print(f"Dimension of TENSOR is {TENSOR.ndim} and shape is {TENSOR.shape}")
print(TENSOR[0])
```

```python
# Generate random tensor

RANDOM_TENSOR = torch.rand(3,4,7)
print(RANDOM_TENSOR)
print(f"Dimension of RANDOM_TENSOR is {RANDOM_TENSOR.ndim} and shape is {RANDOM_TENSOR.shape}")
```

```python
# create random tensor with similar shape to an image tensor
random_image_tensor = torch.rand(size=(224,224,3))
print(random_image_tensor)
print(f"Dimension of random_image_tensor is {random_image_tensor.ndim} and shape is {random_image_tensor.shape}")
```

```python
# create tensor with zeroes and ones

zero = torch.zeros(size=(3,4))
ones = torch.ones(size=(3,4))
print(zero)
print(ones)
print(zero.dtype)
print(ones.dtype)
```

```python
# create range of tensors
tensor_range = torch.arange(start=0,end=10,step=1)
print(tensor_range)

tensor_range_1 = torch.arange(start=20,end=40,step=5)
print(tensor_range_1)
print(f"Data type: {tensor_range_1.dtype}")
print(f"Dimension: {tensor_range_1.ndim}")


```

```python
# create tensor like

ten_zeroes = torch.zeros_like(input=tensor_range)
print(ten_zeroes)
```
